## Relationship

Explanation: Execute this Week-3 exploration step.

```python
%python

display(
    spark.sql("""
        SELECT *
        FROM vehicles v
        LEFT ANTI JOIN collisions c
        ON CAST(v.collision_index AS STRING) =
           CAST(c.collision_index AS STRING)
    """)
)
```

## Bronze demo

## Lineage view

Explanation: Execute this Week-3 exploration step.

```python
spark.sql('CREATE OR REPLACE VIEW workspace.default.roadsafe_week03_lineage_demo_view AS SELECT collision_index,* EXCEPT(collision_index) FROM workspace.default.roadsafe_week03_bronze_demo_collisions')
display(spark.sql('SELECT * FROM workspace.default.roadsafe_week03_lineage_demo_view LIMIT 20'))
```
